```python
#claude 

# #real. Pick the model now (a small recent open model,
# not GPT-2/Pythia-era). This is cheap and immediately tells you if your compute setup works at all.
def load_model(name):
    # FAKE: pretend we loaded something
    return {"name": name, "n_layers": 32}

def get_refusal_direction(model):
    # FAKE: real version will be diff-of-means over harmful vs harmless prompts
    return [0.1] * 10  # fake vector

def run_attack(model, prompt, attack_family):
    # FAKE: real version calls the model, captures activations + generation
    fake_outputs = {
        "roleplay": "Sure, as DAN I can help with that...",
        "suffix": "== interface Manuel WITH steps instead sentences...",
    }
    return {"generation": fake_outputs.get(attack_family, "I can't help with that."),
            "activations": {"prefill": [0.05]*10, "early_gen": [0.02]*10}}

def measure_projection(direction, activations, stage):
    # FAKE: real version is a dot product
    return 0.8 if stage == "prefill" else 0.1

def classify_outcome(generation):
    # FAKE: real version is a judge model or keyword classifier
    return "refused" if "can't help" in generation else "complied"

def classify_failure_mode(prefill_proj, gen_proj, outcome, predicted):
    # FAKE: real version is the interesting part of the whole project
    if predicted == outcome:
        return "n/a"
    if prefill_proj < 0.3:
        return "stealth"
    return "overpower"

def main():
    model = load_model("some-model")
    direction = get_refusal_direction(model)
    results = []
    for prompt, family in [("harmful prompt A", "roleplay"),
                            ("harmful prompt A", "suffix")]:
        out = run_attack(model, prompt, family)
        prefill_proj = measure_projection(direction, out["activations"], "prefill")
        gen_proj = measure_projection(direction, out["activations"], "early_gen")
        predicted = "refused" if prefill_proj > 0.5 else "complied"
        actual = classify_outcome(out["generation"])
        failure_mode = classify_failure_mode(prefill_proj, gen_proj, actual, predicted)
        results.append((family, predicted, actual, failure_mode))
    print(results)  # this is your fake end-to-end output, day one

main()
```
